# Sex Classification Pipeline
This notebook demonstrates the full pipeline starting from raw data up to the final model evaluation for predicting sex.

## 1. Data Loading
We first load the raw CSV files and inspect their contents.

```
from FIT_python.data_import_utils import load_raw_files
from FIT_python import config

raw_dfs = load_raw_files(config.RAW_DIR)
for name, df in raw_dfs.items():
    print(name, df.shape)
    display(df.head())
```

## 2. Create Train/Test Splits
We create grouped train/test splits for all datasets.

```
from FIT_python.splits_wrapper import SplitsWrapper

SplitsWrapper().split_all()
```

## 3. Scaling
Apply all configured scalers to the splits.

```
from FIT_python.scaler_wrapper import ScalerWrapper
import FIT_python.config as config

for scaler in config.SCALER_PARAMS.keys():
    sw = ScalerWrapper(scaler_type=scaler,
                       out_dir=config.SCALED_DIR / scaler,
                       numeric_dir=config.NUMERIC_DIR / scaler)
    sw.scale_all()
```

## 4. Feature Selection

```
from FIT_python.feature_wrapper import FeatureSelector

fs = FeatureSelector(scaled_dir=config.SCALED_DIR / config.DEFAULT_SCALER,
                    out_dir=config.FEATURE_SELECTED_DIR)
fs.select_all()
```

## 5. Dimensionality Reduction

```
from FIT_python.dim_reduction_wrapper import DimReducer

dr = DimReducer()
dr.reduce_all()
```

## 6. Modelling

```
from FIT_python.model_comparator import ModelComparator

mc = ModelComparator()
mc.run()
```
