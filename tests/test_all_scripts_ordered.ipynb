## 1. Data Import
Use DataImporter to load, clean labels, and convert numeric features.

```
from FIT_python.data_import_wrapper import DataImporter
from FIT_python.config import RAW_DIR, DEFAULT_TARGETS
# Initialize importer
importer = DataImporter(
    raw_dir=RAW_DIR,
    target_cols=DEFAULT_TARGETS,
    skip_fillna=['trail'],
    label_map={'f':'female','m':'male'}
)
dfs_ready = importer.run()
for name, df in dfs_ready.items():
    print(f"{name}: {df.shape}")
    display(df.head())
```

### Preview Scaled Data

```
from FIT_python.config import PROCESSED_DIR
import pandas as pd
from pathlib import Path
for ds in (PROCESSED_DIR).iterdir():
    if ds.is_dir():
        p = ds / 'train.parquet'
        if p.exists():
            print(f"{ds.name} scaled train:")
            display(pd.read_parquet(p).head())
```

## 5. Transformation to Numeric Arrays
Execute TransformWrapper

```
from FIT_python.transform_wrapper import TransformWrapper
from FIT_python.config import NUMERIC_DIR
import numpy as np
tw = TransformWrapper()
tw.transform_all()
# Verify outputs
for ds in (NUMERIC_DIR).iterdir():
    if ds.is_dir():
        x = np.load(ds / 'X_train.npy')
        y = np.load(ds / 'y_train.npy')
        print(f"{ds.name}: X_train {x.shape}, y_train {y.shape}")
```
