```python
## 6. Test Feature Selection
!python ../scripts/create_feature_selection.py

from pathlib import Path
import pandas as pd

from FIT_python.config import PROCESSED_DIR, FS_DEFAULT_METHODS, FS_TARGET_FEATURE_COUNTS

# Überprüfe, dass für jedes Dataset und jede Methode/Anzahl eine Datei existiert
for ds_folder in Path("data/processed").iterdir():
    fs_dir = ds_folder / "feature_selection"
    if not fs_dir.exists():
        print(f"No feature_selection folder for {ds_folder.name}")
        continue
    print(f"--- {ds_folder.name} ---")
    for method in FS_DEFAULT_METHODS:
        # Für forward-Methoden: _p und alle counts
        if method.startswith("forward"):
            names = ["forward_p"] + [f"{method}_{cnt}f" for cnt in FS_TARGET_FEATURE_COUNTS]
        else:
            names = [f"{method}_{cnt}f" for cnt in FS_TARGET_FEATURE_COUNTS]
        for name in names:
            path = fs_dir / f"{name}.txt"
            status = "FOUND" if path.exists() else "MISSING"
            print(f"{name}: {status}")
    print()
```
